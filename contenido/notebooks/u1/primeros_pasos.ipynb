## 1.2 · Cómo se ejecuta un programa

`p1_bytecode.py`

```python
import dis

def precio_con_iva(precio):
    return precio * 1.21

print(precio_con_iva(100))
dis.dis(precio_con_iva)          # instrucciones que ejecuta Python
```

`p2_tiempo.py`

```python
import timeit

def suma_bucle(n):
    total = 0
    for i in range(n):           # cada vuelta la interpreta Python
        total += i
    return total

n = 1_000_000
t1 = timeit.timeit(lambda: suma_bucle(n), number=10) / 10
t2 = timeit.timeit(lambda: sum(range(n)), number=10) / 10
print(suma_bucle(n) == sum(range(n)))
print(f"bucle en Python: {t1 * 1000:.1f} ms")
print(f"sum() (escrita en C): {t2 * 1000:.1f} ms")
print(f"sum() es {t1 / t2:.1f} veces más rápida")
```

## 1.4 · Variables y tipos de datos

`p4_tipos.py`

```python
edad = 42
precio = 19.95
nombre = "Lázaro"
mayor = edad >= 18
nada = None
for valor in [edad, precio, nombre, mayor, nada]:
    print(repr(valor), "es de tipo", type(valor).__name__)
print(2 ** 100)                  # los enteros no tienen límite
print(0.1 + 0.2)                 # los decimales se guardan en binario
print(0.1 + 0.2 == 0.3)
print(round(0.1 + 0.2, 2) == 0.3)
```

## 1.5 · Operadores y expresiones

`p8_operadores.py`

```python
segundos = 10_000
horas = segundos // 3600         # división entera
resto = segundos % 3600          # resto de la división
minutos, seg = divmod(resto, 60)
print(f"{segundos} s = {horas} h {minutos} min {seg} s")
print(7 / 2, 7 // 2, 7 % 2, 2 ** 10)
print(2 + 3 * 4, (2 + 3) * 4)    # precedencia
x = 5
x += 2                            # x = x + 2
print(x, 3 < x <= 7, x == 7 and x != 8)
a, b = 1, 2
a, b = b, a                       # intercambio
print(a, b)
```

## 1.6 · Entrada y salida de datos

`p6_formato.py`

```python
productos = [("Pan", 1.2, 3), ("Aceite", 8.95, 1), ("Queso", 6.5, 2)]
print(f"{'Producto':<10}{'Precio':>8}{'Uds.':>6}{'Total':>9}")
total = 0
for nombre, precio, uds in productos:
    importe = precio * uds
    total += importe
    print(f"{nombre:<10}{precio:>8.2f}{uds:>6}{importe:>9.2f}")
print(f"{'TOTAL':<24}{total:>9.2f}")
print(f"{1234567.891:,.2f}")      # separador de miles
print(f"{0.2137:.1%}")             # porcentaje
```

## 1.7 · Un programa completo

`p9_iva.py`

```python
IVA = 0.21                        # tipo general del IVA en España

def desglose(total):
    """Separa base e IVA de un precio con IVA incluido."""
    base = total / (1 + IVA)
    return round(base, 2), round(total - base, 2)

for total in [12.10, 99.99, 250]:
    base, cuota = desglose(total)
    print(f"total {total:>7.2f} € = base {base:>7.2f} + "
          f"IVA {cuota:>6.2f}")
```
