## 2.1 · Valores booleanos y operaciones lógicas

`q5_logicas.py`

```python
print(" a      b      a and b  a or b  not a")
for a in [False, True]:
    for b in [False, True]:
        y, o = a and b, a or b
        print(f" {a!s:<6} {b!s:<6} {y!s:<8} {o!s:<7} {not a}")

def aviso():
    print("  (se ha evaluado aviso())")
    return True

print(False and aviso())         # no hace falta evaluar aviso()
print(True or aviso())
print(True and aviso())
```

## 2.2 · Estructuras condicionales

`q1_condiciones.py`

```python
def calificacion(nota):
    if nota < 0 or nota > 10:
        return "nota no válida"
    elif nota < 5:
        return "suspenso"
    elif nota < 7:
        return "aprobado"
    elif nota < 9:
        return "notable"
    else:
        return "sobresaliente"

for nota in [3.5, 5, 6.9, 8.25, 10, 11]:
    print(f"{nota:>5} -> {calificacion(nota)}")
```

`q10_match.py`

```python
def responder(orden):
    match orden.split():
        case ["hola"]:
            return "¡Hola!"
        case ["suma", a, b]:
            return f"{a} + {b} = {int(a) + int(b)}"
        case ["repite", *palabras]:
            return " ".join(palabras)
        case _:
            return f"no entiendo «{orden}»"

for orden in ["hola", "suma 2 40", "repite tres tristes tigres",
              "adiós"]:
    print(responder(orden))
```

## 2.3 · Bucles

`q2_while.py`

```python
n = 27
pasos = 0
maximo = n
while n != 1:                    # se repite mientras se cumpla
    n = n // 2 if n % 2 == 0 else 3 * n + 1
    pasos += 1
    maximo = max(maximo, n)
print(f"27 llega a 1 en {pasos} pasos; el valor más alto es {maximo}")
```

`q3_for.py`

```python
for i in range(1, 6):
    print(f"7 x {i} = {7 * i}")
total = 0
for k in range(1, 101):
    total += k
print("1 + 2 + ... + 100 =", total)
dias = ["lunes", "martes", "miércoles"]
for posicion, dia in enumerate(dias, start=1):
    print(posicion, dia)
```

`q4_primos.py`

```python
def es_primo(n):
    if n < 2:
        return False
    for d in range(2, int(n ** 0.5) + 1):
        if n % d == 0:
            return False         # sale en cuanto encuentra un divisor
    return True

primos = [n for n in range(50) if es_primo(n)]
print(primos)
n = 1000
while True:
    n += 1
    if not es_primo(n):
        continue                 # pasa al siguiente número
    print("primer primo mayor que 1000:", n)
    break                        # termina el bucle
```

## 2.4 · Operaciones a nivel de bits

`q6_bits.py`

```python
LEER, ESCRIBIR, EJECUTAR = 4, 2, 1          # 100, 010, 001
def texto(p):
    return "".join(c if p & bit else "-" for c, bit in
                   [("r", LEER), ("w", ESCRIBIR), ("x", EJECUTAR)])

modo = 0o755                                 # octal: 7, 5, 5
for quien, desp in [("dueño", 6), ("grupo", 3), ("otros", 0)]:
    p = (modo >> desp) & 0b111               # extrae 3 bits
    print(f"{quien:<6} {p} {p:03b} {texto(p)}")
p = LEER | ESCRIBIR                          # activa dos permisos
print(texto(p), texto(p ^ ESCRIBIR), texto(p & ~LEER))
print(5 << 1, 5 >> 1, bin(5), 0b101 == 5)
```

## 2.5 · Listas

`q7_listas.py`

```python
compra = ["pan", "uvas", "queso"]
compra.append("vino")
compra.insert(0, "aceite")
print(compra, len(compra))
print(compra[0], compra[-1], compra[1:3])
compra.remove("uvas")
compra.sort()
print(compra, "pan" in compra)
otra = compra                    # el mismo objeto, con otro nombre
copia = compra.copy()            # un objeto nuevo
compra.append("sal")
print(otra)
print(copia)
```

`q8_matrices.py`

```python
m = [[1, 2, 3],
     [4, 5, 6],
     [7, 8, 9]]
print(m[1][2])                               # fila 1, columna 2
print([sum(fila) for fila in m])             # suma de cada fila
t = [[m[f][c] for f in range(3)] for c in range(3)]
print(t)                                     # traspuesta
mala = [[0] * 3] * 3             # ¡la misma fila tres veces!
mala[0][0] = 1
print(mala)
buena = [[0] * 3 for _ in range(3)]          # tres filas distintas
buena[0][0] = 1
print(buena)
```

## 2.6 · Un ejemplo: la extensión del *Lazarillo*

`q9_tratados.py`

```python
texto = open("lazarillo.txt", encoding="utf-8").read()
partes = texto.split("Tratado ")              # trozos del libro
print("Prólogo:", len(partes[0].split()), "palabras")
mayor, nombre_mayor = 0, ""
for parte in partes[1:]:
    nombre = parte.split()[0]                 # Primero, Segundo...
    n = len(parte.split())
    print(f"{nombre:<9}{n:>6} {'#' * (n // 200)}")
    if n > mayor:
        mayor, nombre_mayor = n, nombre
print("el tratado más largo es el", nombre_mayor)
```
