## 4.1 · Módulos

`s1_modulos.py`

```python
import math
import random
from datetime import date

print(math.sqrt(2), math.pi)
print(math.floor(2.7), math.ceil(2.1), math.gcd(84, 36))
random.seed(1554)                # misma semilla, mismos resultados
print(random.randint(1, 6), random.choice(["oros", "copas"]))
print(date(2026, 11, 27).strftime("%d/%m/%Y"))
print(date(2026, 12, 25) - date(2026, 11, 27))
print(math.__name__, len(dir(math)), "nombres en math")
```

## 4.2 · Paquetes

`s2_paquete.py`

```python
import textos                    # la carpeta textos/ es un paquete
from textos.limpieza import SIGNOS

libro = open("lazarillo.txt", encoding="utf-8").read()
lista = textos.palabras(libro)
conteo = textos.frecuencias(lista)
print(len(lista), "palabras;", len(SIGNOS), "signos eliminados")
print(textos.mas_frecuentes(conteo, n=8, minimo=5))
print(textos.palabras.__module__, textos.__doc__)
```

## 4.3 · Métodos de cadenas y listas

`s4_cadenas.py`

```python
frase = ("  Pues sepa V.M. ante todas cosas que a mí llaman "
         "Lázaro de Tormes, hijo de Tomé González  ")
f = frase.strip()                # quita espacios de los extremos
print(f.upper()[:30])
print(f.lower().count("a"), f.find("Lázaro"), f.startswith("Pues"))
print(f.replace("V.M.", "Vuestra Merced"))
partes = f.split(", ")
print(partes)
print(" | ".join(partes))
print(f[:4], f[-7:], f[::-1][:7])
print("1554".isdigit(), "Lázaro".isalpha(), "tormes".title())
print(f"[{'Lazarillo':^15}]")
```

`s5_listas.py`

```python
amos = ["ciego", "clérigo", "escudero", "fraile", "buldero"]
print(sorted(amos), sorted(amos, key=len))
amos.append("capellán")
amos.extend(["alguacil", "arcipreste"])
ultimo = amos.pop()
print(ultimo, len(amos), amos.index("escudero"))
print([a.upper() for a in amos if len(a) > 6])
tratados = range(1, 8)
for t, a in zip(tratados, amos):
    print(t, a, end="; ")
print()
```

## 4.4 · Excepciones

`s6_traza.py`

```python
import traceback

def media(valores):
    return sum(valores) / len(valores)

def informe(datos):
    return f"media: {media(datos):.2f}"

try:
    print(informe([]))
except ZeroDivisionError:
    ruta = globals().get("__file__", "s6_traza.py")
    print(traceback.format_exc().replace(ruta, "s6_traza.py"))
```

`s3_excepciones.py`

```python
def precio_unitario(total, unidades):
    if unidades <= 0:
        raise ValueError("las unidades deben ser positivas")
    return total / unidades

for total, uds in [(10, 4), (10, 0), ("10", 2), (10, -1)]:
    try:
        r = precio_unitario(total, uds)
    except ValueError as e:
        print(f"{total!r}, {uds}: valor no válido ({e})")
    except TypeError as e:
        print(f"{total!r}, {uds}: tipo no válido ({e})")
    else:
        print(f"{total!r}, {uds}: {r}")
    finally:
        pass                     # se ejecuta siempre
```
