## 3.1 · Funciones

`r1_funciones.py`

```python
def estadisticas(numeros):
    """Devuelve el mínimo, el máximo y la media de una lista."""
    media = sum(numeros) / len(numeros)
    return min(numeros), max(numeros), media   # una tupla

minimo, maximo, media = estadisticas([4, 8, 15, 16, 23, 42])
print(minimo, maximo, round(media, 2))
print(estadisticas.__doc__)
resultado = estadisticas([1, 2])
print(resultado, type(resultado).__name__)
```

`r2_argumentos.py`

```python
def saludo(nombre, saludo="Hola", final="."):
    return f"{saludo}, {nombre}{final}"

print(saludo("Lázaro"))                       # valores por defecto
print(saludo("Lázaro", "Buenas"))             # por posición
print(saludo("Lázaro", final="!"))            # por nombre

def suma(*numeros):                           # cualquier cantidad
    return sum(numeros)
print(suma(1, 2), suma(1, 2, 3, 4))

def ficha(**datos):                           # pares nombre=valor
    return ", ".join(f"{k}: {v}" for k, v in datos.items())
print(ficha(amo="ciego", lugar="Salamanca"))

def añade_malo(x, lista=[]):                  # ¡error frecuente!
    lista.append(x)
    return lista
print(añade_malo(1), añade_malo(2))
```

`r3_ambito.py`

```python
iva = 0.21                       # variable global

def con_iva(precio):
    total = precio * (1 + iva)   # lee la global; total es local
    return round(total, 2)

print(con_iva(100))
try:
    print(total)
except NameError as e:
    print("Error:", e)

contador = 0
def cuenta():
    global contador              # para modificar la global
    contador += 1
cuenta(); cuenta()
print(contador)
```

`r8_pruebas.py`

```python
def calificacion(nota):
    """Calificación del sistema español para una nota de 0 a 10."""
    if not 0 <= nota <= 10:
        raise ValueError("la nota debe estar entre 0 y 10")
    for limite, nombre in [(5, "suspenso"), (7, "aprobado"),
                           (9, "notable")]:
        if nota < limite:
            return nombre
    return "sobresaliente"

# pruebas: cada assert falla si su condición es falsa
assert calificacion(0) == "suspenso"
assert calificacion(4.99) == "suspenso"
assert calificacion(5) == "aprobado"
assert calificacion(9) == "sobresaliente"
assert calificacion(10) == "sobresaliente"
try:
    calificacion(11)
    assert False, "debería haber fallado"
except ValueError:
    pass
print("todas las pruebas superadas")
```

## 3.2 · Tuplas

`r4_tuplas.py`

```python
punto = (28.1, -15.4)             # latitud y longitud
lat, lon = punto                  # desempaquetado
print(lat, lon, len(punto))
try:
    punto[0] = 0
except TypeError as e:
    print("Error:", e)
ciudades = {(40.4, -3.7): "Madrid", (28.1, -15.4): "Las Palmas"}
print(ciudades[punto])            # una tupla puede ser clave
uno = (5,)                        # tupla de un elemento: con coma
print(type(uno).__name__, type((5)).__name__)
for nombre, edad in [("Ana", 31), ("Luis", 45)]:
    print(nombre, edad)
```

## 3.3 · Diccionarios

`r5_diccionarios.py`

```python
amo = {"tratado": 1, "nombre": "ciego", "lugar": "Salamanca"}
print(amo["nombre"])
print(amo.get("edad"), amo.get("edad", "desconocida"))
amo["lugar"] = "Escalona"         # modificar
amo["paga"] = 0                    # añadir
del amo["tratado"]                 # borrar
for clave, valor in amo.items():
    print(f"{clave:>7}: {valor}")
conteo = {}
for letra in "lazarillo":
    conteo[letra] = conteo.get(letra, 0) + 1
print(conteo)
```

## 3.4 · Procesamiento de datos

`r6_palabras.py`

```python
texto = open("lazarillo.txt", encoding="utf-8").read().lower()
for signo in ".,;:¿?¡!\"'()-«»":
    texto = texto.replace(signo, " ")
palabras = texto.split()
frecuencia = {}
for p in palabras:
    frecuencia[p] = frecuencia.get(p, 0) + 1
print(len(palabras), "palabras,", len(frecuencia), "distintas")
orden = sorted(frecuencia.items(), key=lambda par: par[1],
               reverse=True)
print(orden[:10])
una_vez = [p for p, n in frecuencia.items() if n == 1]
print(len(una_vez), "aparecen una sola vez")
```

`r7_amos.py`

```python
def palabras(texto):
    """Lista de palabras en minúsculas, sin signos."""
    texto = texto.lower()
    for signo in ".,;:¿?¡!\"'()-«»":
        texto = texto.replace(signo, " ")
    return texto.split()

libro = open("lazarillo.txt", encoding="utf-8").read()
tratados = libro.split("Tratado ")[1:]     # con mayúscula
amos = ["ciego", "clérigo", "escudero", "fraile", "buldero",
        "capellán", "alguacil", "arcipreste"]
tabla = {}                         # diccionario de diccionarios
for numero, t in enumerate(tratados, start=1):
    lista = palabras(t)
    tabla[numero] = {a: lista.count(a) for a in amos}
print("tratado " + " ".join(f"{a[:6]:>6}" for a in amos))
for numero, cuenta in tabla.items():
    fila = " ".join(f"{cuenta[a]:>6}" for a in amos)
    print(f"{numero:>7} {fila}")
```
